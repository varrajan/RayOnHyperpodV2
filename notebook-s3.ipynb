## Step 5: Prepare the Dataset

The dataset is written to **S3**. Each Ray Train worker downloads it to its local disk before training.
All paths derive from `s3_base_uri` in `args-s3.yaml`.

````python
import os
import yaml
from datasets import load_dataset

with open("args-s3.yaml") as f:
    CONFIG = yaml.safe_load(f)

S3_BASE_URI = CONFIG["s3_base_uri"].rstrip("/")
MODEL_ID = CONFIG["script_args"]["model_id"]
DATASET_ID = "FreedomIntelligence/medical-o1-reasoning-SFT"
NUM_SAMPLES = 1000

print(f"S3 base URI: {S3_BASE_URI}")

dataset = load_dataset(DATASET_ID, "en", split=f"train[:{NUM_SAMPLES}]")
print(f"Dataset loaded: {len(dataset)} samples")
print(f"Columns: {dataset.column_names}")
print(f"\nExample:\n{dataset[0]}")
````

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

SYSTEM_PROMPT = """You are a medical reasoning assistant. Respond to user queries
with clear, step-by-step medical reasoning."""


def format_chat(example):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": example["Question"]},
        {
            "role": "assistant",
            "content": f"<think>\n{example['Complex_CoT']}\n</think>\n\n{example['Response']}",
        },
    ]
    example["text"] = tokenizer.apply_chat_template(messages, tokenize=False)
    return example


dataset = dataset.map(format_chat)
dataset = dataset.train_test_split(test_size=0.1, seed=42)

train_dataset = dataset["train"]
val_dataset = dataset["test"]

print(f"Train: {len(train_dataset)}, Val: {len(val_dataset)}")
````

````python
import boto3
from urllib.parse import urlparse

LOCAL_DATA = "/tmp/qwen3-medical/datasets"
os.makedirs(f"{LOCAL_DATA}/train", exist_ok=True)
os.makedirs(f"{LOCAL_DATA}/val", exist_ok=True)

# Keep only the formatted chat text — that's what SFTTrainer consumes (dataset_text_field="text").
train_dataset.select_columns(["text"]).to_json(f"{LOCAL_DATA}/train/train.json")
val_dataset.select_columns(["text"]).to_json(f"{LOCAL_DATA}/val/val.json")

_u = urlparse(S3_BASE_URI)
S3_BUCKET, S3_PREFIX = _u.netloc, _u.path.lstrip("/")
s3 = boto3.client("s3")
for split in ("train", "val"):
    key = f"{S3_PREFIX}/datasets/{split}/{split}.json"
    s3.upload_file(f"{LOCAL_DATA}/{split}/{split}.json", S3_BUCKET, key)
    print(f"Uploaded s3://{S3_BUCKET}/{key}")
````

### Resilience

The training script uses `FailureConfig(max_failures=3)`. If a node fails:
1. HyperPod automatically replaces the faulty node
2. Ray reschedules worker pods onto the new node
3. Training resumes from the last checkpoint (persisted to S3 via `RunConfig.storage_path`)

No changes to your training code are needed for this to work.

## Step 10: Verify Output

Model artifacts are in S3 under `<s3_base_uri>/output/`. Ray Train checkpoints are under `<s3_base_uri>/ray_results/`.

````python
def list_s3(prefix):
    total, files = 0, 0
    for page in s3.get_paginator("list_objects_v2").paginate(Bucket=S3_BUCKET, Prefix=prefix):
        for obj in page.get("Contents", []):
            total += obj["Size"]; files += 1
    return files, total

for sub in ("output/final_model", "output/merged_model", "ray_results"):
    n, size = list_s3(f"{S3_PREFIX}/{sub}/")
    status = f"{n} files, {size / 1e9:.2f} GB" if n else "not found — job may still be running"
    print(f"  s3://{S3_BUCKET}/{S3_PREFIX}/{sub}/  ->  {status}")
````

````python
from transformers import pipeline

MERGED_S3_PREFIX = f"{S3_PREFIX}/output/merged_model/"
MERGED_PATH = "/tmp/qwen3-medical/merged_model"

n, _ = list_s3(MERGED_S3_PREFIX)
if n:
    print(f"Downloading merged model ({n} files) to {MERGED_PATH} ...")
    for page in s3.get_paginator("list_objects_v2").paginate(Bucket=S3_BUCKET, Prefix=MERGED_S3_PREFIX):
        for obj in page.get("Contents", []):
            dest = os.path.join(MERGED_PATH, obj["Key"][len(MERGED_S3_PREFIX):])
            os.makedirs(os.path.dirname(dest), exist_ok=True)
            s3.download_file(S3_BUCKET, obj["Key"], dest)

    # The space has no GPU; a 0.6B model generates on CPU in about a minute.
    pipe = pipeline("text-generation", model=MERGED_PATH, torch_dtype="auto", device_map="auto")

    test_prompt = "What are the common symptoms and diagnostic criteria for Type 2 Diabetes?"
    result = pipe([{"role": "user", "content": test_prompt}], max_new_tokens=256)
    print(f"\nPrompt: {test_prompt}")
    print(f"\nResponse:\n{result[0]['generated_text'][-1]['content']}")
else:
    print(f"Merged model not found at s3://{S3_BUCKET}/{MERGED_S3_PREFIX}")
    print("Check that merge_weights is enabled in args-s3.yaml and the job completed.")
````

## Step 12: Deploy with Ray Serve

Deploy the fine-tuned model as a live HTTP endpoint using [Ray Serve](https://docs.ray.io/en/latest/serve/index.html).
The deployment runs on the same Ray cluster, so no additional infrastructure is needed.

The Serve application:
1. Downloads the merged model from S3 to the GPU worker's local disk (same Pod Identity credentials as training)
2. Loads it on **1 GPU** with `transformers` in bf16; the other GPUs on the worker stay free for training
3. Applies the **same system prompt and chat template used in training**, and returns the reasoning
   (`<think>…</think>`) and the final answer separately
4. Serves requests two ways: a Python **DeploymentHandle** from this notebook, and **HTTP** on port 8000

> **Networking:** the Serve HTTP proxy listens on `127.0.0.1` unless told otherwise, and the Ray head
> service created by Studio does not expose port 8000. The cell below starts Serve on `0.0.0.0`
> and finds a reachable URL: the optional `<cluster>-serve-svc` Kubernetes Service (see the access
> table below), else the head pod IP.
>
> **Alternative:** Studio also supports creating a **RayService** from the Tasks tab for a fully
> managed Serve deployment with health checks and automatic rollouts.
>
> For higher throughput, swap the transformers replica for vLLM via `ray.serve.llm` (needs the
> `vllm` package, which is not in the SageMaker Distribution image).

````python
import os
import time

import ray
import requests
from ray import serve
from starlette.requests import Request

# Same reason as Step 4: keep transformers away from the image's TensorFlow/Keras 3
os.environ["USE_TF"] = "0"
os.environ["TRANSFORMERS_NO_TF"] = "1"

ray.init(address="auto", ignore_reinit_error=True)

SERVE_APP_NAME = "medical-qa"
SERVE_ROUTE = "/medical-qa"
SERVE_MODEL_URI = f"s3://{S3_BUCKET}/{S3_PREFIX}/output/merged_model/"

# Must match the system prompt used to format the training data (Step 5)
SERVE_SYSTEM_PROMPT = """You are a medical reasoning assistant. Respond to user queries
with clear, step-by-step medical reasoning."""


@serve.deployment(
    name="MedicalQA",
    num_replicas=1,
    max_ongoing_requests=4,
    ray_actor_options={
        "num_gpus": 1,
        "num_cpus": 4,
        "runtime_env": {
            "env_vars": {
                "USE_TF": "0",
                "TRANSFORMERS_NO_TF": "1",
                # run sync methods (generate) in a thread pool instead of the asyncio loop
                "RAY_SERVE_RUN_SYNC_IN_THREADPOOL": "1",
            }
        },
    },
)
class MedicalQA:
    def __init__(self, model_s3_uri: str, system_prompt: str):
        import boto3
        import torch
        from urllib.parse import urlparse
        from transformers import AutoModelForCausalLM, AutoTokenizer

        # 1. Download the merged model from S3 to the replica's local disk
        parsed = urlparse(model_s3_uri)
        bucket, prefix = parsed.netloc, parsed.path.lstrip("/").rstrip("/") + "/"
        local_path = "/tmp/qwen3-medical/serve_model"
        s3_client = boto3.client("s3")
        for page in s3_client.get_paginator("list_objects_v2").paginate(Bucket=bucket, Prefix=prefix):
            for obj in page.get("Contents", []):
                rel = obj["Key"][len(prefix):]
                if not rel or rel.endswith("/"):
                    continue
                dest = os.path.join(local_path, rel)
                os.makedirs(os.path.dirname(dest), exist_ok=True)
                s3_client.download_file(bucket, obj["Key"], dest)
        print(f"Model downloaded from {model_s3_uri}")

        # 2. Load on the GPU that Ray assigned to this replica
        self.tokenizer = AutoTokenizer.from_pretrained(local_path)
        self.model = AutoModelForCausalLM.from_pretrained(
            local_path, dtype=torch.bfloat16, attn_implementation="sdpa"
        ).to("cuda").eval()
        self.system_prompt = system_prompt
        self.device_name = torch.cuda.get_device_name(0)
        print(f"Model loaded on {self.device_name}, ready for inference")

    def generate(self, prompt: str, max_tokens: int = 1024, temperature: float = 0.6,
                 top_p: float = 0.95) -> dict:
        import torch

        messages = [
            {"role": "system", "content": self.system_prompt},
            {"role": "user", "content": prompt},
        ]
        text = self.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = self.tokenizer(text, return_tensors="pt").to("cuda")

        start = time.time()
        with torch.inference_mode():
            out = self.model.generate(
                **inputs,
                max_new_tokens=max_tokens,
                do_sample=temperature > 0,
                temperature=temperature if temperature > 0 else None,
                top_p=top_p if temperature > 0 else None,
                pad_token_id=self.tokenizer.eos_token_id,
            )
        new_tokens = out[0][inputs["input_ids"].shape[1]:]
        completion = self.tokenizer.decode(new_tokens, skip_special_tokens=True).strip()

        # The model was trained to answer as "<think>reasoning</think> answer"
        truncated = False
        if "</think>" in completion:
            thinking, response = completion.split("</think>", 1)
            thinking, response = thinking.replace("<think>", "").strip(), response.strip()
        elif completion.startswith("<think>"):
            # max_tokens reached while still reasoning: no final answer yet
            thinking, response, truncated = completion.replace("<think>", "").strip(), "", True
        else:
            thinking, response = "", completion

        return {
            "response": response,
            "thinking": thinking,
            "truncated": truncated,
            "new_tokens": int(new_tokens.shape[0]),
            "latency_s": round(time.time() - start, 2),
            "device": self.device_name,
        }

    async def __call__(self, http_request: Request) -> dict:
        import asyncio

        data = await http_request.json()
        # Run the blocking generate() off the event loop so the replica keeps serving health checks
        return await asyncio.to_thread(
            self.generate,
            prompt=data["prompt"],
            max_tokens=int(data.get("max_tokens", 1024)),
            temperature=float(data.get("temperature", 0.6)),
            top_p=float(data.get("top_p", 0.95)),
        )


# Listen on all interfaces so other pods (this workspace included) can reach the proxy.
# If Serve is already running with different HTTP options, Ray keeps the old ones and warns:
# run serve.shutdown() first (this also removes other Serve apps on the cluster).
# serve.run() then logs "The new client HTTP config differs ... ['host'] ... is ignored": that is
# serve.run's own default config (127.0.0.1) being ignored in favor of the 0.0.0.0 set here. Harmless.
serve.start(http_options={"host": "0.0.0.0", "port": 8000})

start = time.time()
handle = serve.run(
    MedicalQA.bind(model_s3_uri=SERVE_MODEL_URI, system_prompt=SERVE_SYSTEM_PROMPT),
    name=SERVE_APP_NAME,
    route_prefix=SERVE_ROUTE,
)
print(f"Serve app '{SERVE_APP_NAME}' {serve.status().applications[SERVE_APP_NAME].status} "
      f"({time.time() - start:.0f}s to deploy)")


def resolve_serve_base_url() -> str:
    """Return the first reachable Serve proxy: the optional serve Service, else the head pod IP."""
    head_ip = next(n["NodeManagerAddress"] for n in ray.nodes()
                   if n["Alive"] and "node:__internal_head__" in n["Resources"])
    candidates = [
        f"http://{RAY_CLUSTER_NAME}-serve-svc.{RAY_NAMESPACE}.svc.cluster.local:8000",
        f"http://{head_ip}:8000",
    ]
    for base in candidates:
        try:
            if requests.get(f"{base}/-/healthz", timeout=3).ok:
                return base
        except requests.RequestException:
            pass
    raise RuntimeError(f"Serve proxy not reachable at any of {candidates}")


SERVE_URL = resolve_serve_base_url() + SERVE_ROUTE
print(f"Deployment live at: {SERVE_URL}")
print("View in Ray Dashboard -> Serve tab for deployment status, replicas, and request metrics")
````

- **Horizontal**: increase `num_replicas` in `@serve.deployment` (each replica takes 1 GPU) to spread load across GPUs and workers
- **Model parallelism**: for models too large for one GPU, set `ray_actor_options={"num_gpus": N}` and load with `device_map="auto"` to shard across the GPUs of one replica
- Ray Serve handles load balancing and replica health checks automatically
- The **Ray Serve Grafana dashboard** (provisioned by the Observability add-on) shows request latency, throughput, and replica status

### Resilience Features (automatic)

| Layer | What it does |
|-------|-------------|
| **Node recovery** | HyperPod replaces faulty nodes; Ray reschedules pods; `FailureConfig` retries the job |
| **Hung job detection** | Per-node Job Monitoring Agent detects stalls; alerts via CloudWatch and Grafana |
| **Tiered checkpointing** | `amzn-sagemaker-checkpointing` writes to local disk first, async uploads to S3 |
| **Serve health checks** | Ray Serve monitors replica health; unhealthy replicas are restarted automatically |

### Key Packages

| Package | Purpose |
|---------|--------|
| `toolkit-for-ray-on-sagemaker-ai` | Remote job submission (`sagemaker_ray://` address), JumpStart model loader, custom hung job detection rules |
| `amzn-sagemaker-checkpointing` | Tiered checkpointing integration (local disk → S3) |
| SageMaker Distribution image | Default container with Ray pre-installed (AWS-maintained) |
